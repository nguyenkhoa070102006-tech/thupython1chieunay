### Bài 3 : Số nguyên tố

```python
from math import *
def nto(n):
    if n < 2 : 
        return False
    else:
        for i in range(2, isqrt(n)+ 1):
            if n % i == 0 :
                return False
        return True
f1 = open("28tech_number.txt" , "r" , encoding = 'utf-8')
f2 = open("28tech_prime.txt" , "w" , encoding = 'utf-8')
a = list(map(int , f1.read().split()))
b = []
for x in a :
    if nto(x):
        b.append(x)
b.sort()
for x in b : 
    f2.write(str(x) + ' ')
f1.close()
f2.close()
```
