```python
class Solution:
    def isHappy(self, n: int) -> bool:
        def next(n):
            res = 0
            while n > 0:
                digit = n % 10
                res += digit**2
                n //= 10   # sửa ở đây
            return res

        seen = set()
        def isHappy(n, seen):
            if n == 1:
                return True
            if n in seen:
                return False
            seen.add(n)
            m = next(n)
            return isHappy(m, seen)

        return isHappy(n, seen)

if __name__ == "__main__":
    sol = Solution()
    print(sol.isHappy(19))
```
