```python
#duyet tung cap so trong list va in ra so lon nhat
num_list = [1,5,7,3,2,8]
k = 2
sub_list = []
result = []
for element in num_list:
    sub_list.append(element)

    if len(sub_list) == k:
        print(sub_list)
        result.append(max(sub_list))
        del sub_list[0]
print(result)
```

```python
#duyet = chi muc
num_list = [3,-1,-2,-1,5,4]
k = 2
start_indexs = list(range(0, len(num_list) - k + 1))
end_indexs = list(range(k, len(num_list) + 1))
result = []
for start_index, end_index in zip(start_indexs, end_indexs):
    sub_list = num_list[start_index: end_index]
    print(sub_list)
    result.append(max(sub_list))
print(result)
```

```python
# đếm số lần xuất hiện chữ cái trong string
character_statistic = {}
word = "baby"
word = word.lower()
for character in word:
    if character in character_statistic:
        character_statistic[character]+=1
    else:
        character_statistic[character]=1
print(character_statistic)
```

```python
source = "hola"
target = "hello"

def edit_distances(source, target):
    distances = []
    for idx in range(len(source) + 1):
        row = [0] * (len(target) + 1)
        distances.append(row)
    print(distances)

    for idx in range(len(source) + 1):
        distances[idx][0] = idx
    for jdx in range(len(target) + 1):
        distances[0][jdx] = jdx

    del_cost = 0
    ins_cost = 0
    sub_cost = 0
    for idx in range(1, len(source) + 1):
        for jdx in range(1, len(target) + 1):
            if source[idx - 1] == target[jdx - 1]:
                distances[idx][jdx] = distances[idx - 1][jdx - 1]
            else:
                del_cost = distances[idx - 1][jdx]
                ins_cost = distances[idx][jdx - 1]
                sub_cost = distances[idx - 1][jdx - 1]
                distances[idx][jdx] = min(del_cost, ins_cost, sub_cost) + 1

    edit_distances = distances[len(source)][len(target)]
    print(distances)
    return edit_distances

print(edit_distances(source, target))
```
