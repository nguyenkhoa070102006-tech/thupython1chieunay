```python
# Câu 3 
import numpy as np
A = np.random.randint(1, 101, 20)
print(np.mean(A), A.std())
B = A[A % 5 == 0]
print(" Phần tử của mảng B là các số chia hết cho 5 :" , B)
print(" Tổng các phần tử của B : " , B.sum())
C = A.reshape(4,5)
print(" Ma trận C : ", C)
tongcot = C.sum(axis = 0)
print(tongcot)
```

### Câu 2.2

```python
import pandas as pd
df = pd.DataFrame({
    "Product": ['A','B','C','D','E'],
    "Category":[ "Food" ,"Tech","Food" , "Tech" , "Food"],
    "Price"  : [ 120 , None , 95 , 340 , 80 ]})
df["Price"] = df["Price"].fillna(df["Price"].median()) # median() là số nắm giữa dãy đc sx tang haoc giảm
print(df.Price)
df["Cheap"] = df["Price"] < 150
print(df.groupby("Category") ["Price"].sum())
print(df["Cheap"].sum())
```

```python
class Circle:
    pi = 3.14
    def __init__(self , radius):
        self.radius = radius 
    def area(self):
        return Circle.pi * self.radius **2
    def grow(self,delta):
        self.radius += delta
    def getpi( self):
        return Circle.pi
c1 = Circle(5)
c1.grow(2)
Circle.pi = 3.14149
print(c1.getpi())
print(round(c1.area(),2))
print(c1.radius > Circle(3).radius)
```
